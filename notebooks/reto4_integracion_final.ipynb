```python
# RETO 4: INTEGRACIÓN FINAL - CALIDAD DEL AGUA
# 👤 Estudiante: Julián Esteban Moscoso Martínez
# 📌 Rol: Unir todo el análisis y dar diagnóstico final

import pandas as pd

# Cargamos los datos (ruta correcta)
datos = pd.read_csv('../data/4_resultados_agua.csv')

print("✅ Archivo cargado. Total registros:", datos.shape[0])
print("📋 Columnas:", datos.columns.tolist())
datos.head()

# --------------------------
# CLASIFICACIÓN DE CALIDAD
# --------------------------
def clasificar_calidad(fila):
    ph = fila['pH']
    oxigeno = fila['oxigeno_disueto']
    temperatura = fila['temperatura']
    
    if 6.0 <= ph <= 9.0 and oxigeno > 5 and temperatura < 30:
        return "✅ ÓPTIMA - CUMPLE NORMATIVA"
    elif (5.0 <= ph < 6.0) or (9.0 < ph <= 10) or (3 <= oxigeno <= 5):
        return "⚠️ ALERTA - RIESGO MODERADO"
    else:
        return "❌ CRÍTICO - NO CUMPLE / CONTAMINACIÓN"

datos['calidad_agua'] = datos.apply(clasificar_calidad, axis=1)
datos[['sitio','pH','oxigeno_disueto','temperatura','calidad_agua']].head(10)

# --------------------------
# RESUMEN ESTADÍSTICO (Reto3)
# --------------------------
datos.describe().round(2)

# --------------------------
# ANÁLISIS POR SITIO (Agrupación)
# --------------------------
analisis_por_sitio = datos.groupby('sitio').agg(
    promedio_ph = ('pH','mean'),
    promedio_oxigeno = ('oxigeno_disueto','mean'),
    promedio_temperatura = ('temperatura','mean'),
    max_turbidez = ('turbidez','max'),
    total_registros = ('sitio','count')
).round(2)

analisis_por_sitio.sort_values('promedio_oxigeno', ascending=False)

# --------------------------
# DIAGNÓSTICO FINAL
# --------------------------
conteo_final = datos['calidad_agua'].value_counts()
porcentajes = (conteo_final / len(datos)*100).round(1)

print("📊 DIAGNÓSTICO GENERAL")
print(conteo_final)
print("\n📌 PORCENTAJES (%)")
print(porcentajes)
```
